```text
Employee Name
```

Tax Slabs (New Tax Regime FY 2025-26):

```text
Up to ₹4,00,000         -> 0%

₹8,00,001 - ₹12,00,000 -> 10%
₹12,00,001 - ₹16,00,000 -> 15%
₹16,00,001 - ₹20,00,000 -> 20%

Above ₹24,00,000       -> 30%
```

Calculate:

```text
Tax Amount = Annual Salary × Tax Rate
Take Home Salary = Annual Salary - Tax Amount
```



### Sample Inputs

```text
Name          : Lakshmi Prasanna
Annual Salary : 820000
```

```text
Income Tax Report
```
